```python
# ==============================================================================
# CODE BLOCK: MOUNT DRIVE VÀ GIẢI NÉN DATASET (CHẠY TRÊN COLAB)
# ==============================================================================
#MUST RUN THIS BLOCK TO GET DATASET IF YOU ARE USING COLAB
from google.colab import drive
import os
import zipfile

# 1. Treo Google Drive vào môi trường Colab
print("🔗 Đang kết nối tới Google Drive...")
drive.mount('/content/drive')

# 2. Định nghĩa đường dẫn
# Giả sử file zip của cậu đang nằm ngoài cùng của Drive (My Drive)
# Cậu có thể sửa lại tên folder nếu lưu ở chỗ khác (ví dụ: '/content/drive/MyDrive/Bách_Khoa/captcha_dataset.zip')
zip_path_in_drive = '/content/drive/MyDrive/Captcha_Reader/captcha_dataset.zip'
local_extract_path = './' # Giải nén ngay tại thư mục làm việc hiện tại của Colab

# 3. Tiến hành giải nén cực tốc bằng Python (hoặc lệnh bash)
if os.path.exists(zip_path_in_drive):
    print(f"📦 Đã tìm thấy file zip tại Drive. Đang giải nén vào {local_extract_path}...")

    with zipfile.ZipFile(zip_path_in_drive, 'r') as zip_ref:
        zip_ref.extractall(local_extract_path)

    print("✅ Giải nén thành công!")
    # Kiểm tra xem cấu trúc thư mục xuất hiện đúng chưa
    if os.path.exists('./captcha_dataset'):
        print(f"📂 Thư mục dữ liệu đã sẵn sàng: {os.listdir('./captcha_dataset')}")
    else:
        print("⚠️ Cảnh báo: Giải nén xong nhưng không thấy thư mục 'captcha_dataset', kiểm tra lại cấu trúc file zip nhé cậu.")
else:
    print(f"❌ LỖI: Không tìm thấy file zip tại đường dẫn: {zip_path_in_drive}")
    print("👉 Cậu hãy check lại xem đã vứt file 'captcha_dataset.zip' vào đúng thư mục My Drive chưa nhé!")
```
